```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("Social_Network_Ads.csv")
dataset
```

```python
dataset=(pd.get_dummies(dataset,drop_first=True,dtype=int))
```

```python
dataset=dataset.drop(columns='User ID', axis=1) # axis=0 is for rows 
```

```python
dataset
```

```python
dataset["Purchased"].value_counts()
```

```python
dataset.columns
```

```python
independent = dataset[['Age', 'EstimatedSalary','Gender_Male']]
```

```python
independent.shape
```

```python
independent
```

```python
dependent = dataset[['Purchased']]
```

```python
dependent
```

```python
dependent.shape
```

```python
from sklearn.model_selection import train_test_split #Split Test and training data
x_train,x_test,y_train,y_test = train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler #standarizing the value if we have abnormal variance
sc=StandardScaler()
x_train=sc.fit_transform(x_train) #Mean value and Standard deviation Value got calculated
x_test=sc.transform(x_test)
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
param_grid = [
    {
        "solver": ["lbfgs", "newton-cg", "sag", "liblinear", "saga"],
        "penalty": ["l2"],
        "C": np.logspace(-3, 3, 7)  
    }
]
log_reg = LogisticRegression(max_iter=1000)
grid_search = GridSearchCV(estimator=log_reg, param_grid=param_grid, refit = "true", verbose=3,  cv=5, scoring="f1_weighted", n_jobs=-1)
grid_search = grid_search.fit(x_train,y_train)
```

```python
re=grid_search.cv_results_
grid_prediction = grid_search.predict(x_test) #Predicting the test data
print("The R_score value for the best parameter {}:".format(grid_search.best_params_))
```

```python
from sklearn.metrics import confusion_matrix #Calculating confusion matrix
cm = confusion_matrix(y_test,grid_prediction)
from sklearn.metrics import classification_report
clf_report = classification_report(y_test,grid_prediction)
```

```python
#F1score - we are seperately evaluating the f1 score value, however we will be able to see in the clf-report
from sklearn.metrics import f1_score
f1_macro=f1_score(y_test,grid_prediction,average='weighted')
print("The f1_score value for the best parameter {}:".format(grid_search.best_params_))
```

```python
print("The Confudion Matrix:", cm)
```

```python
print("The clf report:", clf_report)
```
